# Cleaning the data for the Quebec Border shapefile 

```python
import geopandas as gpd 
import numpy as np 
import pandas as pd
import json

#Using the line shape file
file = gpd.read_file("../site/data/SDA.shp", layer='regio_l')
```

## Projecting the data 
Just like for the fire, the borders data needs to be in the correct projection for MapLibre  

```python
# Checking the CRS
print(file.crs)
```

```python
# Save GeoJSON in WGS84 coordinates, as required by MapLibre GeoJSON sources
file = file.to_crs(epsg=4632)
```

## Understanding the Data  

```python
print(file.columns.tolist())

print(file["RES_CO_TYP"].unique())
```

## Getting rid of unnessary rows and columns 
Reducing the file size but getting rid of all the unneeded columns: Everything other than geometry

```python

rows_to_drop = file.index[file["RES_CO_TYP"].isin(["LIMREG", "FRO52"])]
file = file.drop(rows_to_drop)

file = file.drop(
    columns=["RES_ID_IND", "RES_CO_TYP"]
)

file.simplify(0.1).to_file("../site/data/clean_SDA.geojson", driver="GeoJSON")

file = gpd.read_file("../site/data/clean_SDA.geojson")
```
