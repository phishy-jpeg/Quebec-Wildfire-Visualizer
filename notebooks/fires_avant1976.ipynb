# Cleaning the Data for data for firest before 1976
Source: https://www.donneesquebec.ca/recherche/dataset/feux-de-foret

```python
# imports
import geopandas as gpd
import os
import numpy as np
import json

# Loading the data
file = gpd.read_file("../site/data/FEUX_ANCIENS_PROV.gpkg", layer="FEUX_ANCIENS_PROV")
```

## Understanding the data
Getting to know the columns, amount of rows, and general information about the data

```python
print('Column names: ', file.columns)
print('first five rows of data: ', file.head())
print('Num rows in the file: ', len(file))
```

```python
# Dropping un needed columns
file_filtered = file.drop(
    columns=['geoc_fan',
             'shape_length',
             'shape_area']
)
# Changing the crs
file_filtered = file_filtered.to_crs(epsg=3857)  

# Saving the new updated file 
file_filtered.simplify(0.01)
file_filtered.to_file("../site/data/filtered_feux_ancien.gpkg")
```
