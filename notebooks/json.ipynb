# Creating JSON files of the amount of fires per year
In this file I am taking the fire data from feux_prov and feux_ancien and creating a JSON file, containing the amount of fires per year, and the amount of hectars burned per year.

```python
import geopandas as gpd 
import numpy as np 
import pandas as pd
import json

# Reading the necesary files 
feux_ancien = gpd.read_file("../site/data/FEUX_ANCIENS_PROV.gpkg", layer="FEUX_ANCIENS_PROV")
feux = gpd.read_file("../site/data/FEUX_PROV.gpkg", layer="FEUX_PROV")
```

```python
year_list = {}

for index, row in feux_ancien.iterrows():
    key = row['exercice']
    if key in year_list:
        year_list[key]['fires'] += 1
        year_list[key]['hectares'] += row['superficie']

    elif key not in year_list:
        year_list[key] = {'fires': 1, 'hectares': row.get('superficie')}

for index, row in feux.iterrows():
    key = row['exercice']
    if key in year_list:
        year_list[key]['fires'] += 1
        year_list[key]['hectares'] += row['superficie']

    elif key not in year_list:
        year_list[key] = {'fires': 1, 'hectares': row['superficie']}
```

```python
# Saving the data to a json file 
with open('../site/data/fire_counts_by_year.json', 'w') as f:
    json.dump(year_list, f)
```
